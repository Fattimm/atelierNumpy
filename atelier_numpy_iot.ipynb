## Partie 1:

### Importation du module Numpy

```python
import numpy as np
```

### Création du listes python stockant ces températures

```python
temperature_listes = [24.5, 25.1, 26.3, 27.0, 25.8]
print(temperature_listes)
```

### Création du tableau NumPy équivalent stockant ces températures 

```python
temperature_listes_np = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
print(temperature_listes_np)
```

###  Multiplication par 2 sur ces deux collections

```python
resultat_listes = temperature_listes * 2
print(resultat_listes)

resultat_listes_np = temperature_listes_np * 2
print(resultat_listes_np)
```

###  Différence que l'on observe

- Avec la **liste Python**, l'opérateur `*` duplique la liste : les 5 éléments sont répétés deux fois, ce qui donne une liste de 10 éléments.
- Avec le **tableau NumPy**, l'opérateur `*` effectue une multiplication mathématique élément par élément : chaque température est multipliée par 2, ce qui donne un tableau de 5 éléments (les valeurs doublées).

Cela montre que NumPy permet des opérations **vectorisées** (calcul mathématique direct sur toutes les valeurs), contrairement aux listes Python qui traitent les opérateurs de façon littérale (concaténation/répétition).

## Partie 8 : Calcul scientifique 

#### Considérons le tableau représentant les 24 températures du 1er bâtiment. 
#### 1) Pour les températures calculer le minimum, le maximum, la somme, la moyenne, la médiane, la variance et l’écart-type 

```python
minimum = np.min(temperature_jour)
maximum = np.max(temperature_jour)
somme = np.sum(temperature_jour)
moyenne = np.mean(temperature_jour)
mediane = np.median(temperature_jour)
variance = np.var(temperature_jour)
ecart_type = np.std(temperature_jour)

print("Min :", minimum)
print("Max :", maximum)
print("Somme :", somme)
print("Moyenne :", moyenne)
print("Médiane :", mediane)
print("Variance :", variance)
print("Écart-type :", ecart_type)
```

#### 2) Standardiser le tableau des températures puis vérifier les valeurs de la moyenne et de l’écarttype du tableau standardisé

```python
temperature_standardisees = (temperature_jour - moyenne) / ecart_type
print(temperature_standardisees)

print("Nouvelle moyenne :", np.mean(temperature_standardisees))
print("Nouvel écart-type :", np.std(temperature_standardisees))
```

#### 3) Normaliser (Min-Max) le tableau des températures puis vérifier les valeurs du tableau normalisé

```python
temperature_normalisees = (temperature_jour - minimum) / (maximum - minimum)
print(temperature_normalisees)

print("Min du tableau normalisé :", np.min(temperature_normalisees))
print("Max du tableau normalisé :", np.max(temperature_normalisees))
```

#### 4) Convertir le tableau initial en Fahrenheit 

```python
temperature_fahrenheit = temperature_jour * 9/5 + 32
print(temperature_fahrenheit)
```
