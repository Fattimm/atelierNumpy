## Partie 1:

### Importation du module Numpy

```python
import numpy as np
```

### Création du listes python stockant ces températures

```python
temperature_listes = [24.5, 25.1, 26.3, 27.0, 25.8]
print(temperature_listes)
```

### Création du tableau NumPy équivalent stockant ces températures 

```python
temperature_listes_np = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
print(temperature_listes_np)
```

###  Multiplication par 2 sur ces deux collections

```python
resultat_listes = temperature_listes * 2
print(resultat_listes)

resultat_listes_np = temperature_listes_np * 2
print(resultat_listes_np)
```

###  Différence que l'on observe

- Avec la **liste Python**, l'opérateur `*` duplique la liste : les 5 éléments sont répétés deux fois, ce qui donne une liste de 10 éléments.
- Avec le **tableau NumPy**, l'opérateur `*` effectue une multiplication mathématique élément par élément : chaque température est multipliée par 2, ce qui donne un tableau de 5 éléments (les valeurs doublées).

Cela montre que NumPy permet des opérations **vectorisées** (calcul mathématique direct sur toutes les valeurs), contrairement aux listes Python qui traitent les opérateurs de façon littérale (concaténation/répétition).

## Partie 4 – Indexation et slicing

#### Considérant le tableau représentant les 24 températures d'une journée, afficher :
